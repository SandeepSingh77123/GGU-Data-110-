```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
```

```python
plt.subplot(2,2,1)
plt.plot(y,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

```python
fig = plt.figure(figsize=(12,8),dpi = 100)
axis1 = fig.add_axes([0.1,0.1,0.4,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
```
